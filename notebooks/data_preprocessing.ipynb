```python
import pandas as pd
import numpy as np
from pathlib import Path

# Identify the project root
PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

# Set the dataset path
DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "synthetic"
    / "platelet_inventory_demand.csv"
)

print("Project root:", PROJECT_ROOT)
print("Dataset path:", DATA_PATH)
print("Dataset exists:", DATA_PATH.exists())

# Load the dataset
if not DATA_PATH.exists():
    raise FileNotFoundError(f"Dataset not found: {DATA_PATH}")

df = pd.read_csv(DATA_PATH)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes.to_frame(name="Data Type"))

print("\nFirst five records:")
display(df.head())

print("\nDataset information:")
df.info()
```

```python
print("=" * 45)
print("OPERATIONAL DATASET VALIDATION")
print("=" * 45)

print("Total rows:", len(df))
print("Total columns:", len(df.columns))
print("Unique hospitals:", df["hospital_id"].nunique())
print("Unique blood groups:", df["blood_group"].nunique())
print("Missing cells:", df.isna().sum().sum())

print(
    "Duplicate records:",
    df.duplicated(
        subset=["hospital_id", "date", "blood_group"]
    ).sum()
)

print(
    "All records marked synthetic:",
    df["is_synthetic"].astype(bool).all()
)
```

```python
expected_closing = (
    df["opening_inventory"]
    + df["incoming_units"]
    - df["units_used"]
    - df["units_expired"]
)

print(
    "Inventory balance errors:",
    (expected_closing != df["closing_inventory"]).sum()
)

print(
    "Negative closing inventory:",
    (df["closing_inventory"] < 0).sum()
)

print(
    "Usage exceeds demand:",
    (df["units_used"] > df["daily_demand"]).sum()
)

print(
    "Unmet-demand errors:",
    (
        df["unmet_demand"]
        != df["daily_demand"] - df["units_used"]
    ).sum()
)
```

```python
# Step 9.2: Check time coverage

df["date"] = pd.to_datetime(df["date"], errors="coerce")

print("Earliest date:", df["date"].min())
print("Latest date:", df["date"].max())

print(
    "Missing or invalid dates:",
    df["date"].isna().sum()
)

print(
    "Duplicate hospital/date/blood-group records:",
    df.duplicated(
        subset=["hospital_id", "date", "blood_group"]
    ).sum()
)
```
